```python
# Importing Required Libraries
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
# Making DataFrame
data = {
    'Mileage': [10000,20000,30000,40000,50000,
                 60000,70000,80000,90000,100000],

    'Price': [12,11,10,9,8,7,6.5,5.8,5,4.5]
}
df = pd.DataFrame(data)
X=df[['Mileage']]
y=df['Price']
```

```python
# Visualize the Data
X = df[['Mileage']]
y = df['Price']

plt.scatter(df['Mileage'], df['Price'])

plt.xlabel("Mileage")
plt.ylabel("Price")

plt.title("Mileage vs Price")

plt.show()
```

```python
# Splitting the Data
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.3,random_state=42
)
# Training Data (70%) and Test Data(30%)
```

```python
# Fitting and Training the model
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
# Predicting the Value
new_mileage = [[82000]]
predicted_price = model.predict(new_mileage)
print(f"Predicted Price :{predicted_price}")
```

```python
# Import Libraries For Evaluating the Model
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

# Predict the value for test data
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("R2 Score:", r2)
```
