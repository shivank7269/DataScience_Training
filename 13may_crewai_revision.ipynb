```python
import os
import pandas as pd
from crewai import Agent, Task, Crew
from dotenv import load_dotenv

# =====================================================
# LOAD ENV VARIABLES
# =====================================================

load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

# =====================================================
# SAMPLE TRAFFIC DATA
# =====================================================

traffic_data = pd.DataFrame({
    "Location": [
        "Junction A",
        "Junction B",
        "Highway 1",
        "Sector 18",
        "City Center"
    ],
    "Vehicle_Count": [320, 150, 520, 410, 280],
    "Average_Speed": [18, 42, 25, 15, 30],
    "Signal_Status": ["Red", "Green", "Green", "Red", "Yellow"]
})

# =====================================================
# TRAFFIC SUMMARY
# =====================================================

average_speed = traffic_data["Average_Speed"].mean()

most_congested = traffic_data.loc[
    traffic_data["Vehicle_Count"].idxmax()
]["Location"]

congestion_percentage = (
    len(traffic_data[traffic_data["Average_Speed"] < 20])
    / len(traffic_data)
) * 100

traffic_summary = f"""
Traffic Dataset:

{traffic_data.to_string(index=False)}

Key Metrics:
- Average Speed: {average_speed:.2f} km/h
- Most Congested Area: {most_congested}
- Congestion Percentage: {congestion_percentage:.2f}%
"""

# =====================================================
# AGENT 1: TRAFFIC DATA ANALYST
# =====================================================

traffic_analyst = Agent(
    role="Traffic Data Analyst",
    goal="Analyze traffic data and identify congestion patterns",
    backstory=(
        "You are an urban traffic expert skilled in congestion detection, "
        "traffic KPI calculation, and smart city analytics."
    ),
    llm="groq/llama-3.3-70b-versatile",
    verbose=True
)

# =====================================================
# AGENT 2: TRAFFIC OPTIMIZATION STRATEGIST
# =====================================================

traffic_strategist = Agent(
    role="Traffic Optimization Strategist",
    goal="Suggest intelligent traffic optimization strategies",
    backstory=(
        "You are a smart traffic strategist who improves traffic flow, "
        "optimizes signals, and designs emergency vehicle routing systems."
    ),
    llm="groq/llama-3.3-70b-versatile",
    verbose=True
)

# =====================================================
# AGENT 3: SMART CITY REPORT GENERATOR
# =====================================================

report_generator = Agent(
    role="Smart City Report Generator",
    goal="Generate professional traffic management reports",
    backstory=(
        "You are a government traffic reporting specialist who creates "
        "executive-level smart city traffic reports."
    ),
    llm="groq/llama-3.3-70b-versatile",
    verbose=True
)

# =====================================================
# TASK 1: TRAFFIC ANALYSIS
# =====================================================

analysis_task = Task(
    description=f"""
    Analyze the following traffic dataset and identify:

    - Congested traffic zones
    - Peak traffic areas
    - Traffic KPIs
    - Traffic-related problems

    Traffic Data:
    {traffic_summary}
    """,
    expected_output=(
        "Detailed traffic analysis including congestion hotspots, "
        "KPIs, and traffic problems"
    ),
    agent=traffic_analyst
)

# =====================================================
# TASK 2: TRAFFIC OPTIMIZATION
# =====================================================

strategy_task = Task(
    description="""
    Based on the traffic analysis:

    Suggest:
    - Better signal timing strategies
    - Alternate traffic routes
    - Emergency vehicle handling plans
    - Smart traffic optimization recommendations
    """,
    expected_output=(
        "Traffic optimization strategies and smart routing recommendations"
    ),
    agent=traffic_strategist,
    context=[analysis_task]
)

# =====================================================
# TASK 3: SMART CITY REPORT
# =====================================================

report_task = Task(
    description="""
    Generate a professional Smart City Traffic Report.

    Include:
    - Executive summary
    - Traffic KPIs
    - Congested zones
    - Optimization recommendations
    - Emergency vehicle solutions
    """,
    expected_output="Final professional traffic management report",
    agent=report_generator,
    context=[analysis_task, strategy_task]
)

# =====================================================
# CREW SETUP
# =====================================================

crew = Crew(
    agents=[
        traffic_analyst,
        traffic_strategist,
        report_generator
    ],
    tasks=[
        analysis_task,
        strategy_task,
        report_task
    ],
    verbose=True,
    tracing=True
)

# =====================================================
# RUN SYSTEM
# =====================================================

result = crew.kickoff()

# =====================================================
# FINAL OUTPUT
# =====================================================

print("\n🚦 SMART TRAFFIC MANAGEMENT REPORT 🚦\n")

print(result)
```
